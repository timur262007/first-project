```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/sm_small_onlyMostFrequent.csv')
```

```python
df['product_category'].value_counts()
```

```python
sns.countplot(df, x="product_category", hue="product_is_delivery")
plt.title('Доставляемость разных типов продуктов')
```

```python
grouped = df.groupby('product_category')['product_price'].mean().reset_index()
```

```python

grouped.plot.bar(x='product_category', y='product_price', title='Столбчатая диаграмма по средней цене товара')
```

```python
grouped1 = df.groupby('product_is_pick_up', as_index=False).agg({'product_price' : 'mean'}).sort_values('product_price', ascending=False)
```

```python
mean_price = df.groupby('product_is_pick_up')['product_price'].mean()
```

```python
grouped1.plot.bar(x = 'product_is_pick_up', y = 'product_price', title = 'средняя цена по заказам доступным для самовывоза')
```

```python
cut_table = df.iloc[:, 2:].drop(['product_category', 'product_gender', 'product_age', 'create_timestamp'], axis=1)
```

```python
cut_table
```

```python
final_table = cut_table.apply(pd.to_numeric, errors='coerce').dropna()
```

```python
final_table1 = final_table.corr()[['product_price']]
final_table1.rename(columns = {'product_price':'corr_with_price'})
```

```python
final_table1.plot.bar()
plt.title('корреляция цены товара по сравнению с другими параметрами')
```

```python
df_1 = df.groupby('device_type', as_index=True).agg({'user_id' : 'count'})
df_1.rename(columns={'user_id' : 'counts_users'})
df_1.plot.bar()
```

```python
df_2 = df.groupby(['user_id'], as_index=False).agg({'product_price':'mean'}).rename(columns={'product_price':'mean_price'})
```
